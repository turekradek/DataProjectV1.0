```python
import pyspark
from pyspark.sql import SparkSession
```

```python
# Tworzenie sesji Spark (lokalny klaster wewnątrz kontenera)
spark = SparkSession.builder \
    .appName("MojaAnaliza") \
    .getOrCreate()

# Sprawdzenie, czy Spark żyje
print(f"Wersja Spark: {spark.version}")

# Przykładowe dane
data = [("Warszawa", 21.5), ("Kraków", 19.8), ("Berlin", 20.2)]
df = spark.createDataFrame(data, ["Miasto", "Temperatura"])
```
