```python
from pyspark.sql import SparkSession
```

```python

# Inicjalizacja sesji Spark z obsługą sterownika Postgres
# Spark automatycznie pobierze driver, jeśli masz dostęp do internetu w kontenerze
spark = SparkSession.builder \
    .appName("PostgresConnectionTest") \
    .config("spark.jars.packages", "org.postgresql:postgresql:42.7.1") \
    .getOrCreate()

# Konfiguracja połączenia na podstawie Twojego docker-compose
# Używamy nazwy serwisu 'postgresdb' zamiast localhost
jdbc_url = "jdbc:postgresql://postgresdb:5432/postgres"
connection_properties = {
    "user": "postgres",
    "password": "password",  # upewnij się, że to hasło zgadza się z Twoim YAML
    "driver": "org.postgresql.Driver"
}

try:
    # Próba odczytu listy tabel systemowych, aby sprawdzić połączenie
    df = spark.read.jdbc(url=jdbc_url, table="information_schema.tables", properties=connection_properties)
    
    print("✅ Sukces! Połączono z Postgres.")
    print("Oto pierwsze 5 tabel znalezionych w bazie:")
    df.select("table_name").show(5)

except Exception as e:
    print("❌ Błąd połączenia!")
    print(e)
```
